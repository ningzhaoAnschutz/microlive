# Filter Exported Data

Keep only the exported TIF + AVI files whose series were **actually processed** in the GUI (i.e., have a `results_` folder).

The matched files are copied into a new `exported_filtered/` folder, preserving the original subfolder structure.

---

## 1. Define paths for all tags

```python
# # ============================================================
# # Paths for each tag variant
# # ============================================================
base = "/Volumes/LaCie/UTag_paper_data/Harringtonine"

tag_configs = {
    "AlfaTag": {
        "exported_folder": f"{base}/exported_data/AlfaTag",
        "results_folder":  f"{base}/AlfaTag/HT_Analysis_GUI",
    },
    "SunTag": {
        "exported_folder": f"{base}/exported_data/SunTag",
        "results_folder":  f"{base}/SunTag/HT_Analysis_GUI",
    },
    "UTag": {
        "exported_folder": f"{base}/exported_data/UTag",
        "results_folder":  f"{base}/UTag/HT_Analysis_GUI",
    },
    "UTag_CF": {
        "exported_folder": f"{base}/exported_data/UTag_CF",
        "results_folder":  f"{base}/UTag_CF/HT_Analysis_GUI",
    },
}
# # ============================================================
```

## Dry run — preview what will be kept

```python
import sys
from pathlib import Path
from export_lif_data import filter_exported_data

# Add export module to path
export_dir = Path.cwd() if Path("export_lif_data.py").exists() else Path(__file__).parent
if str(export_dir) not in sys.path:
    sys.path.insert(0, str(export_dir))

for tag_name, cfg in tag_configs.items():
    print(f"\n{'='*60}")
    print(f"  {tag_name}")
    print(f"{'='*60}")
    kept, skipped = filter_exported_data(
        exported_folder=cfg["exported_folder"],
        results_folder=cfg["results_folder"],
        output_folder=f"{base}/exported_filtered/{tag_name}",
        dry_run=True,
    )
    print(f"  → {tag_name}: {kept} would be kept, {skipped} skipped")
```

## Run the filter (copy files)

```python
for tag_name, cfg in tag_configs.items():
    print(f"\n{'='*60}")
    print(f"  {tag_name}")
    print(f"{'='*60}")
    kept, skipped = filter_exported_data(
        exported_folder=cfg["exported_folder"],
        results_folder=cfg["results_folder"],
        output_folder=f"{base}/exported_filtered/{tag_name}",
        dry_run=False,
    )
    print(f"  ✓ {tag_name}: {kept} copied, {skipped} skipped")

print("\n✓ All tags filtered.")
```
