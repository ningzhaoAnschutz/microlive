# LIF Batch Export

Export **OME-TIFF** and **AVI** (with timestamp + scale bar) for every series in every `.lif` file found in the input folder.

**Output structure:**
```
<data_folder>/exported_data/
    ├── <lif_name>/
    │   ├── <lif_name>_<series_1>.ome.tif
    │   ├── <lif_name>_<series_1>.avi
    │   ├── <lif_name>_<series_2>.ome.tif
    │   └── <lif_name>_<series_2>.avi
```

---

## Set the data folder path

````python
# ============================================================
# USER: Set the path to the folder containing your .lif files
# ============================================================
# data_folder = "/Volumes/LaCie/UTag_paper_data/Puro"
#data_folder = "/Volumes/LaCie/UTag_paper_data/ACF"
data_folder = "/Volumes/LaCie/UTag_paper_data/Harringtonine"

# ============================================================
````

## Run the export

````python
import sys
from pathlib import Path
from export_lif_data import batch_export

# Add export module to path
export_dir = Path.cwd() if Path("export_lif_data.py").exists() else Path(__file__).parent
if str(export_dir) not in sys.path:
    sys.path.insert(0, str(export_dir))

# Validate input folder
data_path = Path(data_folder)
assert data_path.is_dir(), f"Folder not found: {data_path}"

# Create output folder
output_folder = data_path / "exported_data"
output_folder.mkdir(parents=True, exist_ok=True)

# List .lif files found
lif_files = sorted(data_path.glob("**/*.lif"))
print(f"Data folder   : {data_path}")
print(f"Output folder : {output_folder}")
print(f"LIF files     : {len(lif_files)}")
for f in lif_files:
    print(f"  • {f.relative_to(data_path)}")

# Run batch export: TIF + AVI (timestamp + scale bar)
batch_export(
    input_folder=str(data_path),
    output_folder=str(output_folder),
    export_tif=True,
    export_video=False,
    show_timestamp=True,
    show_scalebar=True,
    fps=10,
    channel=0,
    max_projection= True,
)

print("\n✓ Export complete.")
````
